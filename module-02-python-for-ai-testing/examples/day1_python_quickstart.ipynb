## 1. Variables & Data Types

A variable is a named container for a value. Python figures out the type automatically — no declarations needed.

### Core types you'll use constantly

| Type | Example | Use case in AI testing |
|---|---|---|
| `str` | `"Hello"` | Prompts, model responses, labels |
| `int` | `42` | Token counts, retry counts, indices |
| `float` | `0.7` | Temperature, similarity scores, thresholds |
| `bool` | `True` | Pass/fail flags, feature toggles |
| `None` | `None` | Missing values, unset optional params |
| `list` | `["a", "b"]` | Batches of prompts, list of test cases |
| `dict` | `{"k": "v"}` | API payloads, test case records, configs |

> **Plain English:** a variable is a label you stick on a box. You can change what's in the box any time. Python reads the content of the box to decide what type it is — you don't have to declare it.

```python
# The five scalar types
model_name   = "gpt-4o-mini"   # str
max_tokens   = 512              # int
temperature  = 0.7              # float
is_streaming = False            # bool
api_key      = None             # None — not set yet

print(type(model_name))     # <class 'str'>
print(type(temperature))    # <class 'float'>
print(type(is_streaming))   # <class 'bool'>
```

```python
# isinstance() — the right way to check type in a test
print(isinstance(temperature, float))       # True
print(isinstance(max_tokens, (int, float))) # True — either is fine
print(isinstance(model_name, int))          # False

# Never use type(x) == float directly — isinstance handles subclasses
print(isinstance(True, int))  # True! bool is a subclass of int in Python
```

```python
# Type conversion — common patterns when parsing LLM output
raw_score = "0.87"              # model returned a string
score = float(raw_score)        # convert to float for math
print(score > 0.5)              # True

raw_count = "42"
token_count = int(raw_count)
print(f"{token_count} tokens used")

# Be careful: int() on a float string fails
try:
    int("0.87")   # ValueError!
except ValueError as e:
    print(f"Error: {e}  → use float() first, then int()")
    print(int(float("0.87")))   # 0 — truncates, doesn't round
```

```python
# Truthiness — Python's flexible truth system
# These are all falsy:
print(bool(""))          # empty string → False
print(bool(0))           # zero → False
print(bool(None))        # None → False
print(bool([]))          # empty list → False
print(bool({}))          # empty dict → False

# These are all truthy:
print(bool("hello"))     # any non-empty string → True
print(bool(42))          # any non-zero number → True
print(bool([1, 2]))      # non-empty list → True

# Practical: check if a response has content
response = ""
if not response:
    print("Model returned an empty response — fail!")
else:
    print(f"Got {len(response)} chars")
```

```python
# Comparison operators
score = 0.72

print(score > 0.7)     # True
print(score >= 0.7)    # True
print(score == 0.72)   # True  — but avoid == on floats!
print(score != 0.5)    # True

# Safe float comparison
from math import isclose
print(isclose(score, 0.72, rel_tol=1e-9))  # True

# Chaining comparisons (Python feature, very readable)
print(0.5 < score < 0.9)   # True — score is between 0.5 and 0.9
```

```python
# Boolean operators: and, or, not
has_content = True
no_toxicity = True
has_keywords = False

# All conditions must pass
passed = has_content and no_toxicity and has_keywords
print(f"All checks: {passed}")   # False — keyword check failed

# At least one must pass
any_issue = not no_toxicity or not has_content
print(f"Any issue: {any_issue}")  # False — both are fine

# Short-circuit evaluation: Python stops at the first False in `and`
# This is why the order of checks matters for performance
```

```python
def addition(a: int,b: int=10) -> int:
    if a>b :
        return a - b
    if b>a:
        return b - a
    if a==b:
        return a+b
```

```python
# Basic function anatomy
def classify_score(score: float) -> str:
    """Return a quality label for a 0.0–1.0 evaluation score."""
    if score >= 0.9:
        return "excellent"
    if score >= 0.7:
        return "good"
    if score >= 0.5:
        return "acceptable"
    return "poor"

print(classify_score(0.95))   # excellent
print(classify_score(0.65))   # acceptable
print(classify_score(0.3))    # poor

# The type hints `: float` and `-> str` are not enforced at runtime
# but they make the function self-documenting and enable IDE autocompletion
```

```python
# Default arguments — parameters with a fallback value
def build_prompt(
    question: str,
    persona: str = "a helpful assistant",
    language: str = "English",
    max_words: int = 100,
) -> str:
    """Assemble a structured prompt from components."""
    return (
        f"You are {persona}. Answer in {language}, "
        f"in {max_words} words or fewer.\n\nQuestion: {question}"
    )

# Call with only the required argument
print(build_prompt("What is RAG?"))
print()
# Override defaults
print(build_prompt("Was ist RAG?", language="German", max_words=50))
```

```python
# Multiple return values (returned as a tuple)
def evaluate_response(response: str, expected_keyword: str) -> tuple[bool, str]:
    """Return (passed, reason)."""
    if not response:
        return False, "empty response"
    if expected_keyword.lower() not in response.lower():
        return False, f"missing keyword: '{expected_keyword}'"
    if len(response) > 2000:
        return False, f"response too long: {len(response)} chars"
    return True, "all checks passed"

passed, reason = evaluate_response("The capital of France is Paris.", "Paris")
print(f"passed={passed}  reason={reason}")

passed, reason = evaluate_response("I don't know.", "Paris")
print(f"passed={passed}  reason={reason}")
```

```python
# *args — accept any number of positional arguments
def all_keywords_present(response: str, *keywords: str) -> bool:
    """Return True only if every keyword appears in the response."""
    response_lower = response.lower()
    return all(kw.lower() in response_lower for kw in keywords)

response = "The transformer architecture uses self-attention and positional encoding."
print(all_keywords_present(response, "transformer", "attention"))  # True
print(all_keywords_present(response, "transformer", "LSTM"))       # False
```

```python
def multi_add(*args : int,**kwargs: int) -> int:
    total = sum(args)
    for k, v in kwargs.items():
        total += v
    return total
```

```python
# **kwargs — accept any number of keyword arguments
def make_api_payload(prompt: str, **kwargs) -> dict:
    """Build an API request body. Extra kwargs override defaults."""
    payload = {
        "model": "gpt-4o-mini",
        "temperature": 0.3,
        "max_tokens": 500,
        "messages": [{"role": "user", "content": prompt}],
    }
    payload.update(kwargs)   # override with any extras passed in
    return payload

print(make_api_payload("Hello"))
print()
print(make_api_payload("Hello", temperature=0.9, max_tokens=100))
```

```python
# sorted(), min(), max() with a key function
test_results = [
    {"id": "q1", "score": 0.95, "latency_ms": 1200},
    {"id": "q2", "score": 0.45, "latency_ms": 800},
    {"id": "q3", "score": 0.82, "latency_ms": 2100},
]

# Sort by score descending
by_score = sorted(test_results, key=lambda r: r["score"], reverse=True)
for r in by_score:
    print(f"  {r['id']}: score={r['score']}")

# Worst performer
worst = min(test_results, key=lambda r: r["score"])
print(f"Worst: {worst['id']} ({worst['score']})")

# Slowest
slowest = max(test_results, key=lambda r: r["latency_ms"])
print(f"Slowest: {slowest['id']} ({slowest['latency_ms']}ms)")
```

```python
# Regular expressions — essential for parsing LLM output
import re

response = "The answer is 42. Confidence: 87.5%. Source: Wikipedia."

# Find all numbers (int or float)
numbers = re.findall(r'\d+\.?\d*', response)
print(numbers)   # ['42', '87.5']

# Extract a percentage
match = re.search(r'Confidence: (\d+\.?\d*)%', response)
if match:
    confidence = float(match.group(1))
    print(f"Confidence: {confidence}%")

# Check format of a response
date_pattern = r'\d{4}-\d{2}-\d{2}'
has_date = bool(re.search(date_pattern, "Meeting on 2025-01-15"))
print(f"Has date: {has_date}")
```

## 9. The `__main__` Guard

Every Python file is both a **script** (can be run directly) and a **module** (can be imported).
The `if __name__ == "__main__":` guard ensures setup code only runs when the file is executed directly — not when imported by another script or by pytest.

> **Why this matters in testing:** if your test utilities run side effects on import (print, API calls, file writes), pytest will execute them before your tests even start. The main guard prevents this.

```python
# Example of good module structure:
"""
# my_llm_client.py

ANTHROPIC_API_KEY = os.environ.get("ANTHROPIC_API_KEY")

def generate(prompt: str) -> str:
    # ... API call ...
    pass

def run_sample_prompts():
    prompts = ["Hello", "What is 2+2?"]
    for p in prompts:
        print(generate(p))

# This only runs when you do: python my_llm_client.py
# NOT when you do: from my_llm_client import generate
if __name__ == "__main__":
    run_sample_prompts()
"""
print("The __main__ guard is what separates a script from a reusable module.")
print(f"This notebook's __name__ is: {__name__}")
```
